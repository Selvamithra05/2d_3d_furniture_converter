```python
import torch
import numpy as np
import cv2
import trimesh
from PIL import Image
from google.colab import files
import matplotlib.pyplot as plt
import os
from transformers import DPTImageProcessor, DPTForDepthEstimation

device = "cuda" if torch.cuda.is_available() else "cpu"

processor = DPTImageProcessor.from_pretrained("Intel/dpt-hybrid-midas")
midas = DPTForDepthEstimation.from_pretrained("Intel/dpt-hybrid-midas")
midas.to(device)
midas.eval()

def get_depth(image):
    inputs = processor(images=image, return_tensors="pt").to(device)
    with torch.no_grad():
        outputs = midas(**inputs)
        depth = outputs.predicted_depth
    depth = torch.nn.functional.interpolate(
        depth.unsqueeze(1),
        size=image.size[::-1],
        mode="bicubic",
        align_corners=False,
    ).squeeze()
    depth = depth.cpu().numpy()
    depth = (depth - depth.min()) / (depth.max() - depth.min())
    return depth

uploaded = files.upload()
image_path = list(uploaded.keys())[0]

image = Image.open(image_path).convert("RGB")

plt.imshow(image)
plt.axis("off")
plt.show()

depth_map = get_depth(image)

plt.imshow(depth_map, cmap="gray")
plt.axis("off")
plt.show()

os.chdir("/content/TripoSR")

input_path = "input.png"
image.save(input_path)

if os.path.exists("run.py"):
    !python run.py --image input.png --output output_mesh.obj
elif os.path.exists("inference.py"):
    !python inference.py --image input.png --output output_mesh.obj
else:
    print("TripoSR script not found")

if os.path.exists("output_mesh.obj"):
    print("3D model generated successfully")
    files.download("output_mesh.obj")
else:
    print("Mesh generation failed")
```

```python
import sys
import torch
import numpy as np
import cv2
from PIL import Image
import trimesh
import os
from tsr.system import TSR
from google.colab import files

sys.path.append("./TripoSR")

uploaded = files.upload()
image_path = list(uploaded.keys())[0]

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading MiDaS depth model...")
midas = torch.hub.load("intel-isl/MiDaS", "MiDaS_small")
midas.to(device)
midas.eval()
transform = torch.hub.load("intel-isl/MiDaS", "transforms").small_transform

def get_depth(image):
    img = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2BGR)
    input_batch = transform(img).to(device)
    with torch.no_grad():
        prediction = midas(input_batch)
        prediction = torch.nn.functional.interpolate(
            prediction.unsqueeze(1),
            size=img.shape[:2],
            mode="bicubic",
            align_corners=False,
        ).squeeze()
    depth = prediction.cpu().numpy()
    depth = (depth - depth.min()) / (depth.max() - depth.min())
    return depth

print("Loading TripoSR model...")
model = TSR.from_pretrained(
    "stabilityai/TripoSR",
    config_name="config.yaml",
    weight_name="model.ckpt",
)

model.to(device)
model = model.float()

torch.cuda.empty_cache()

image = Image.open(image_path).convert("RGB").resize((512,512))

print("Estimating depth...")
depth_map = get_depth(image)

print("Generating REAL 3D mesh from YOUR image...")

with torch.no_grad():
    scene_codes = model([image], device=device)

meshes = model.extract_mesh(scene_codes, has_vertex_color=True)

output_file = "real_3d_model.obj"

trimesh.exchange.export.export_mesh(meshes[0], output_file)

print(f"3D model saved at: {output_file}")

files.download(output_file)
```
